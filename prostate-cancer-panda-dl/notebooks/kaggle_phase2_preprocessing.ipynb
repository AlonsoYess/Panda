# PANDA - Fase 2 Preprocesamiento Inicial

Este notebook ejecuta la Fase 2 sin entrenamiento: splits + extraccion controlada de tiles.

## 1. Configuracion inicial

```python
from pathlib import Path
import subprocess
import os

import yaml
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image
from IPython.display import display

repo_root = Path.cwd()
if not (repo_root / 'config.yaml').exists() and (Path('/kaggle/working/prostate-cancer-panda-dl/config.yaml')).exists():
    repo_root = Path('/kaggle/working/prostate-cancer-panda-dl')

os.chdir(repo_root)
print(f'Repo root: {repo_root}')

config_path = repo_root / 'config.yaml'
with config_path.open('r', encoding='utf-8') as f:
    config = yaml.safe_load(f)

config
```

## 2. Verificacion de rutas

```python
required_paths = {
    'data_root': Path(config['data_root']),
    'train_csv': Path(config['train_csv']),
    'train_images_dir': Path(config['train_images_dir']),
    'train_label_masks_dir': Path(config['train_label_masks_dir']),
    'sample_submission_csv': Path(config['sample_submission_csv']),
}

for key, p in required_paths.items():
    kind = 'DIR' if p.is_dir() else 'FILE'
    exists = p.exists()
    print(f'[{key}] exists={exists} | type={kind} | path={p}')
```

## 3. Crear splits

```python
def run_cmd(cmd: str):
    print(f'Running: {cmd}')
    result = subprocess.run(cmd, shell=True, capture_output=True, text=True)
    print(result.stdout)
    if result.stderr:
        print('STDERR:')
        print(result.stderr)
    print(f'Exit code: {result.returncode}')
    return result

res_split = run_cmd('python scripts/02_create_splits.py')
assert res_split.returncode == 0, 'Fallo la creacion de splits'
```

## 4. Ejecutar extraccion de tiles en modo prueba

```python
print(f"max_slides actual en config: {config.get('max_slides')}")
res_tiles = run_cmd('python scripts/03_extract_tiles.py')
assert res_tiles.returncode == 0, 'Fallo la extraccion de tiles'
```

## 5. Mostrar distribucion de splits

```python
outputs_dir = Path(config['outputs_dir'])
splits_path = outputs_dir / 'metadata' / 'splits.csv'
splits_df = pd.read_csv(splits_path)

print('splits.csv:', splits_path)
display(splits_df.head())
display(splits_df['split'].value_counts())
display(pd.crosstab(splits_df['split'], splits_df['isup_grade']))
display(pd.crosstab(splits_df['split'], splits_df['cancer_label']))
```

## 6. Resumen de tiles candidatos

```python
cand_path = outputs_dir / 'metadata' / 'candidate_tiles_manifest.csv'
cand_df = pd.read_csv(cand_path)

print('candidate_tiles_manifest.csv:', cand_path)
print('Total candidatos:', len(cand_df))
display(cand_df.head())
display(cand_df[['tissue_pct', 'mask_pct']].describe())
display(cand_df['split'].value_counts())
```

## 7. Resumen de tiles seleccionados

```python
tile_path = outputs_dir / 'metadata' / 'tile_manifest.csv'
tile_df = pd.read_csv(tile_path)

print('tile_manifest.csv:', tile_path)
print('Total seleccionados:', len(tile_df))
display(tile_df.head())
display(tile_df['split'].value_counts())
display(tile_df.groupby('slide_id')['tile_id'].count().describe())
```

## 8. Visualizar algunos tiles seleccionados

```python
sample_tiles = tile_df.head(min(6, len(tile_df))).copy()
fig, axes = plt.subplots(2, 3, figsize=(12, 8))
axes = axes.flatten()

for ax in axes:
    ax.axis('off')

for i, (_, row) in enumerate(sample_tiles.iterrows()):
    p = Path(row['tile_path'])
    if p.exists():
        img = Image.open(p).convert('RGB')
        axes[i].imshow(img)
        axes[i].set_title(f"{row['slide_id']}\nmask_pct={row['mask_pct']:.3f}")
        axes[i].axis('off')

plt.tight_layout()
plt.show()
```

## 9. Visualizar relacion tile y mascara (si disponible)

```python
try:
    import openslide  # type: ignore
except Exception:
    openslide = None

if openslide is None:
    print('OpenSlide no disponible en este entorno.')
else:
    row = tile_df[tile_df['mask_available'] == 1].head(1)
    if row.empty:
        print('No hay tiles con mascara disponible en este run.')
    else:
        r = row.iloc[0]
        mask_path = Path(r['mask_path'])
        if not mask_path.exists():
            print('Mask path no existe:', mask_path)
        else:
            mslide = openslide.OpenSlide(str(mask_path))
            level = int(r['level'])
            level = min(level, mslide.level_count - 1)
            downsample = float(mslide.level_downsamples[level])
            x0 = int(round(int(r['x']) * downsample))
            y0 = int(round(int(r['y']) * downsample))
            size = int(r['tile_size'])
            mask_tile = mslide.read_region((x0, y0), level, (size, size)).convert('RGB')
            mslide.close()

            tile_img = Image.open(Path(r['tile_path'])).convert('RGB')

            fig, ax = plt.subplots(1, 2, figsize=(8, 4))
            ax[0].imshow(tile_img)
            ax[0].set_title('Tile seleccionado')
            ax[0].axis('off')

            ax[1].imshow(mask_tile)
            ax[1].set_title('Mascara en misma region')
            ax[1].axis('off')
            plt.tight_layout()
            plt.show()
```
