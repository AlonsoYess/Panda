# PANDA - Validacion de Acceso (Fase 1)

Este notebook valida rutas, CSV, WSI y mascaras del dataset PANDA en Kaggle.

## 1. Configuracion inicial

```python
from pathlib import Path
import random
import os

import numpy as np
import pandas as pd
from IPython.display import display

try:
    import openslide  # type: ignore
except Exception:
    openslide = None

try:
    import tifffile
except Exception:
    tifffile = None

try:
    import yaml
except Exception as ex:
    raise RuntimeError("pyyaml es requerido para este notebook") from ex

def set_seed(seed: int = 42) -> None:
    random.seed(seed)
    np.random.seed(seed)
    os.environ["PYTHONHASHSEED"] = str(seed)

def info(msg: str) -> None:
    print(f"[INFO] {msg}")

def ok(msg: str) -> None:
    print(f"[OK] {msg}")

def warn(msg: str) -> None:
    print(f"[WARN] {msg}")

set_seed(42)
info("Semilla fijada en 42")
```

```python
# Resolver config.yaml de forma robusta
config_candidates = [
    Path("config.yaml"),
    Path("../config.yaml"),
    Path("/kaggle/working/prostate-cancer-panda-dl/config.yaml"),
]

config_path = next((p for p in config_candidates if p.exists()), None)
if config_path is None:
    warn("No se encontro config.yaml. Se usara configuracion por defecto para Kaggle.")
    config = {
        "data_root": "/kaggle/input/prostate-cancer-grade-assessment",
        "train_csv": "train.csv",
        "train_images_dir": "train_images",
        "train_label_masks_dir": "train_label_masks",
        "test_images_dir": "test_images",
        "sample_submission_csv": "sample_submission.csv",
        "outputs_dir": "/kaggle/working/outputs",
    }
else:
    info(f"Cargando config desde: {config_path}")
    with config_path.open("r", encoding="utf-8") as f:
        config = yaml.safe_load(f)

def resolve_path(base_root: Path, value: str) -> Path:
    p = Path(value)
    return p if p.is_absolute() else (base_root / p)

data_root = Path(config["data_root"])
paths = {
    "data_root": data_root,
    "train_csv": resolve_path(data_root, config["train_csv"]),
    "train_images_dir": resolve_path(data_root, config["train_images_dir"]),
    "train_label_masks_dir": resolve_path(data_root, config["train_label_masks_dir"]),
    "test_images_dir": resolve_path(data_root, config["test_images_dir"]),
    "sample_submission_csv": resolve_path(data_root, config["sample_submission_csv"]),
    "outputs_dir": Path(config["outputs_dir"]),
}

paths
```

## 2. Validacion de rutas

```python
required_files = ["train_csv", "sample_submission_csv"]
required_dirs = ["train_images_dir", "train_label_masks_dir", "test_images_dir"]

for key in required_files:
    target = paths[key]
    if target.is_file():
        ok(f"{key} encontrado: {target}")
    else:
        warn(f"{key} NO encontrado: {target}")

for key in required_dirs:
    target = paths[key]
    if target.is_dir():
        ok(f"{key} encontrado: {target}")
    else:
        warn(f"{key} NO encontrado: {target}")
```

## 3. Carga de train.csv

```python
train_df = None
try:
    train_df = pd.read_csv(paths["train_csv"])
    ok("train.csv cargado correctamente")
    info(f"Registros totales: {len(train_df)}")
    info(f"Columnas ({len(train_df.columns)}): {list(train_df.columns)}")
except Exception as ex:
    warn(f"No se pudo cargar train.csv: {ex}")

train_df.head() if train_df is not None else None
```

## 4. Analisis basico de ISUP y Gleason

```python
if train_df is not None:
    if "isup_grade" in train_df.columns:
        info("Distribucion de isup_grade")
        display(train_df["isup_grade"].value_counts(dropna=False).sort_index())
    else:
        warn("No existe columna isup_grade")

    if "gleason_score" in train_df.columns:
        info("Distribucion de gleason_score")
        display(train_df["gleason_score"].value_counts(dropna=False).sort_index())
    else:
        warn("No existe columna gleason_score")
else:
    warn("Analisis omitido: train_df no disponible")
```

## 5. Verificacion de imagenes y mascaras

```python
def count_tiff_files(folder: Path) -> int:
    if not folder.exists():
        return 0
    return sum(1 for p in folder.glob("*.tiff") if p.is_file())

def count_mask_files_with_primary_pattern(folder: Path) -> int:
    if not folder.exists():
        return 0
    return sum(1 for p in folder.glob("*_mask.tiff") if p.is_file())

n_train_images = count_tiff_files(paths["train_images_dir"])
n_train_masks = count_tiff_files(paths["train_label_masks_dir"])
n_train_masks_primary = count_mask_files_with_primary_pattern(paths["train_label_masks_dir"])

info(f".tiff en train_images: {n_train_images}")
info(f".tiff en train_label_masks: {n_train_masks}")
info(f"Mascaras con patron *_mask.tiff: {n_train_masks_primary}")
```

## 6. Lectura de 2-3 WSI de ejemplo

```python
def get_mask_candidates(image_id: str, masks_dir: Path) -> list[Path]:
    return [
        masks_dir / f"{image_id}_mask.tiff",
        masks_dir / f"{image_id}.tiff",
    ]

def resolve_mask_path(image_id: str, masks_dir: Path) -> Path | None:
    for candidate in get_mask_candidates(image_id=image_id, masks_dir=masks_dir):
        if candidate.exists():
            return candidate
    return None

def inspect_wsi(path: Path) -> None:
    if not path.exists():
        warn(f"Archivo no existe: {path.name}")
        return

    info(f"Leyendo archivo: {path.name}")

    if openslide is not None:
        try:
            slide = openslide.OpenSlide(str(path))
            ok(
                "OpenSlide -> "
                f"dimensiones base={slide.dimensions}, niveles={slide.level_count}, "
                f"dimensiones_por_nivel={slide.level_dimensions}"
            )
            slide.close()
            return
        except Exception as ex:
            warn(f"No se pudo leer con OpenSlide ({path.name}): {ex}")
    else:
        warn("OpenSlide no esta disponible. Se intentara con tifffile.")

    if tifffile is not None:
        try:
            with tifffile.TiffFile(path) as tif:
                n_pages = len(tif.pages)
                n_series = len(tif.series)
                first_shape = tif.series[0].shape if n_series > 0 else None
            ok(
                "tifffile -> "
                f"paginas={n_pages}, series={n_series}, forma_primera_serie={first_shape}"
            )
            return
        except Exception as ex:
            warn(f"No se pudo leer con tifffile ({path.name}): {ex}")
    else:
        warn("tifffile no esta disponible para fallback.")

    warn(f"No se pudo extraer metadata de {path.name} con los lectores disponibles.")

def select_examples(df: pd.DataFrame, masks_dir: Path, n: int = 3) -> pd.DataFrame:
    if "image_id" not in df.columns:
        raise ValueError("train.csv no contiene la columna 'image_id'.")

    n = min(n, len(df))
    if n == 0:
        return df.iloc[0:0]

    df_local = df.copy()
    df_local["image_id"] = df_local["image_id"].astype(str)

    mask_names = set()
    if masks_dir.exists():
        mask_names = {path.name for path in masks_dir.glob("*.tiff") if path.is_file()}

    def pick_mask_filename(image_id: str):
        primary = f"{image_id}_mask.tiff"
        fallback = f"{image_id}.tiff"
        if primary in mask_names:
            return primary
        if fallback in mask_names:
            return fallback
        return None

    df_local["_mask_filename"] = df_local["image_id"].apply(pick_mask_filename)
    df_local["_has_mask"] = df_local["_mask_filename"].notna()

    with_mask = df_local[df_local["_has_mask"]]
    without_mask = df_local[~df_local["_has_mask"]]

    selected_frames = []
    n_with_mask = min(n, len(with_mask))
    if n_with_mask > 0:
        selected_frames.append(with_mask.sample(n=n_with_mask, random_state=42))

    remaining = n - n_with_mask
    if remaining > 0 and len(without_mask) > 0:
        selected_frames.append(without_mask.sample(n=remaining, random_state=42))

    if not selected_frames:
        return df_local.iloc[0:0]

    return pd.concat(selected_frames, axis=0).sample(frac=1, random_state=42).reset_index(drop=True)

if train_df is not None and "image_id" in train_df.columns and len(train_df) > 0:
    sample_rows = select_examples(train_df, masks_dir=paths["train_label_masks_dir"], n=3)
    display_cols = [c for c in ["image_id", "isup_grade", "gleason_score", "data_provider"] if c in sample_rows.columns]
    display_cols += [c for c in ["_has_mask", "_mask_filename"] if c in sample_rows.columns]
    display(sample_rows[display_cols])

    for _, row in sample_rows.iterrows():
        image_id = str(row["image_id"])
        image_path = paths["train_images_dir"] / f"{image_id}.tiff"
        mask_candidates = get_mask_candidates(image_id=image_id, masks_dir=paths["train_label_masks_dir"])
        mask_path = resolve_mask_path(image_id=image_id, masks_dir=paths["train_label_masks_dir"])

        info(f"--- Validando image_id={image_id} ---")
        info(f"Ruta imagen WSI: {image_path}")
        info(f"Imagen WSI existe: {'SI' if image_path.exists() else 'NO'}")
        info(f"Ruta mascara principal esperada: {mask_candidates[0]}")
        info(f"Ruta mascara fallback: {mask_candidates[1]}")
        info(f"Mascara existe: {'SI' if mask_path is not None else 'NO'}")

        inspect_wsi(image_path)

        if mask_path is not None:
            info(f"Mascara asociada encontrada. Ruta final: {mask_path}")
            inspect_wsi(mask_path)
        else:
            warn(f"Mascara no encontrada para {image_id} usando ambos patrones esperados.")

else:
    warn("No se pudo ejecutar lectura de WSI (train_df vacio o sin image_id).")
```

## 7. Conclusion de validacion

```python
if train_df is None:
    warn("Validacion incompleta: train.csv no se pudo cargar.")
else:
    ok("Validacion completada: rutas, CSV, WSI y mascaras evaluadas.")
    info("Fase 1 finalizada sin entrenamiento, sin tiles, sin embeddings y sin modelos.")
```
